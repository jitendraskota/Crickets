## Step 1. Load the dataset using **pandas** library -> The key library for data analysis in Python

```python
import pandas as pd
df = pd.read_csv("crickets.csv")
df.head()
```

## Step 2. Create a scatter plot to visualize the relationship between chirp rate and temperature using **matplotlib** library -> The basic visualization library in python

```python
import matplotlib.pyplot as plt

plt.scatter(df["rate"], df["temp"])
plt.xlabel("Chirp rate")
plt.ylabel("Temperature (°F)")
plt.title("Chirp rate vs Temperature")
plt.show()
```

## Step 3. Fit the Least Squares Model to the data using **statsmodels** library -> Used to build statistical models in python
---



```python
import statsmodels.api as sm

x = sm.add_constant(df["rate"])   # adds the intercept term
y = df["temp"]

model = sm.OLS(y, x).fit() #Ordinary Least Squares Regression - Standard and most common method for Linear Regression
print(model.summary())
```

## Step 4. Observe the key numbers from the result

```python
print("Slope:", model.params["rate"])
print("Intercept:", model.params["const"])
print("R-squared:", model.rsquared)
print("P-value for rate:", model.pvalues["rate"])
```

## Step 5. Simple Linear Regression - Include the Regression line and view the plot again

```python
plt.scatter(df["rate"], df["temp"], label="Data")
plt.plot(df["rate"], model.predict(x), color="red", label="Fitted line") #Includes the regression line
plt.xlabel("Chirp rate")
plt.ylabel("Temperature (°F)")
plt.legend()
plt.show()
```

## Step 6. Predict the temperature for a chirp rate not in the dataset

```python
new_rate = pd.DataFrame({"const": [1], "rate": [100]})
predicted_temp = model.predict(new_rate)
print("Predicted temperature:", predicted_temp.values[0])
```

## Step 7. Multiple Linear Regression - Extend the same concept to more than one independent variable

```python
import seaborn as sns # using seaborn library for a better visualization
sns.scatterplot(data=df, x="rate", y="temp", hue="species")
```
